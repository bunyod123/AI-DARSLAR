```python
# 2
class MenXaqimda:
    def __init__(self, ism, familya, viloyat, birthday):
        self.firstname = ism
        self.lastNmae = familya
        self.state = viloyat
        self.tyil = birthday

    def tanishtirsam_Ozimni(self):
        return f"Ismim {self.firstname} familyam {self.lastNmae} {self.state}da tugulganman va {self.tyil} da tugilganman"
    
    def get_name(self):
        return self.firstname
    
    def get_age(self, yil):
        return yil - self.tyil
```

```python
info1 = MenXaqimda('Bunyod','Turgunboev','Namangan',1037)
info2 = MenXaqimda('Rustam','Qodirov','Tashkent',1888)
```

```python
info1.tanishtirsam_Ozimni()
```

```python
info2.tanishtirsam_Ozimni()
```

```python
class Talaba:
    def __init__(self,ism,familya,manzil):
        self.ism = ism
        self.familya = familya
        self.manzil = manzil
        self.baho = 3
        
    def info(self):
        return f"Men {self.ism}man va familyam {self.familya}dur brodar"
    
    def info2(self):
        return f"Meni manzilim {self.manzil}da va bahoyim {self.baho} erur"
```

```python
class BankAccount:
    def __init__(self,current, deposit, withdrawl):
        self.current = current
        self.deposit = deposit
        self.withdrawl = withdrawl
        
    def PulQoshish(self):
        self.current = self.current + self.deposit
        return f"Pul qoshilgandan song xisobda {self.current} boldi"
    
    def PulYechish(self):
        self.current = self.current - self.withdrawl
        return f"Xisobdan {self.withdrawl} yechildi va xisobda {self.current} qoldi"
```

```python
natija = BankAccount(10000, 2500,1200)

print(natija.PulQoshish())

print(natija.PulYechish())
```
