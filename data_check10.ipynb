```python
import pandas as pd
from joblib import dump,load

from sklearn.model_selection import train_test_split,KFold,cross_val_score
from sklearn.preprocessing import LabelEncoder
labeller = LabelEncoder()

from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor 
RF_Reg = RandomForestRegressor()
DT_Reg = DecisionTreeRegressor()

from sklearn.metrics import r2_score,mean_absolute_error
```

```python
import numpy as np
from tabulate import tabulate
```

```python
df = pd.read_csv(r"C:\Users\bunyo\OneDrive\Desktop\git_dataset\9. Sales-Data-Analysis.csv")
```

```python
df.sample(5)
```

```python
df.info()
```

```python
def encodlash(df):
    for cols in df.columns:
        if df[cols].dtype == 'object':
            if df[cols].nunique() < 4:
                dum = pd.get_dummies(df[cols], prefix='new', dtype=int)
                df = pd.concat([df.drop(columns=[cols]), dum], axis=1)
            else:
                df[cols] = labeller.fit_transform(df[cols])
    return df
```

```python
df = encodlash(df)
```

```python
df.info()
```

```python
x = df.drop(columns=['Quantity'],axis=1)
y = df['Quantity']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=34)
```

# Decision tree regression

```python
DT_Reg.fit(x_train,y_train)
```

```python
DT_pred = DT_Reg.predict(x_test)
```

```python
DT_r2 = r2_score(y_test,DT_pred)
print(DT_r2)
```

# Random Forrest Regression

```python
RF_Reg.fit(x_train,y_train)
```

```python
RF_pred = RF_Reg.predict(x_test)
```

```python
RF_r2 = r2_score(y_test,RF_pred)
print(RF_r2)
```

# KFold validation bilan train qilish (Decision Tree)

```python
kf = KFold(n_splits=6, shuffle=True, random_state=32)

cvs = cross_val_score(DT_Reg, X=x, y=y, cv=kf,scoring='r2')

mean_cvs = np.mean(cvs)
```

```python
print(mean_cvs)
```

# KFold validation bilan (Random Forrest)

```python
kf_1 = KFold(n_splits=6, shuffle=True, random_state=33)

cvs_1 = cross_val_score(RF_Reg, X=x, y=y, cv=kf_1, scoring='r2')

mean_cvs_1 = np.mean(cvs_1)
```

```python
print(mean_cvs_1)
```

# Tabulate table

```python
natija = [
    ['Decision tree reg', DT_r2, mean_cvs],
    ['Random forest reg', RF_r2, mean_cvs_1]
]

sarlavha = ['Model name', 'natural score', 'score with KFold']

table = tabulate(natija, headers=sarlavha,tablefmt='fancy_grid',floatfmt='.5f')
```

```python
print(table)
```

```python
def count_by(x, n):
    result = []
    we = 1
    for i in n:
        result.append(x*we)
        we += 1
    return result
```

```python
def func2(x,n):
    result = []
    for i in range(n+1):
        result.append(x*i)
    return result
```
