# datasetni data type larini tekshiramiz

```python
data_type = df.dtypes.value_counts()
data_type
```

```python
plt.figure(figsize=(8,6))
plt.pie(data_type, labels=['object','integer','float'], explode=[0.1,0.1,0.1], autopct='%1.1f%%',shadow=True, startangle=90)
plt.title('Float, object va integer datalar nisbati')
plt.show()
```

# visualizatsiya uchun function yozamiz

```python
def plot_pie_count(column_name, data=df):
    plt.figure(figsize=(12, 5))

    # Pie Plot
    plt.subplot(1, 2, 1)
    data[column_name].value_counts().plot.pie(autopct='%1.1f%%', startangle=90)
    plt.title(f'Distribution of {column_name.capitalize()}', fontsize=16, fontweight='bold', color='black')
    plt.ylabel('')

    # Count Plot
    plt.subplot(1, 2, 2)
    sns.countplot(x=column_name, data=data, palette='viridis')
    plt.title(f'Count of {column_name.capitalize()}', fontsize=16, fontweight='bold', color='black')
    plt.xlabel(column_name.capitalize(), fontsize=12)
    plt.ylabel("Count", fontsize=12)
    plt.xticks(rotation = 10)
    plt.tight_layout()
    plt.show()
```

```python
plot_pie_count('transmission') 
```

```python
plot_pie_count('recommended_fuel')
```

```python
plot_pie_count('drive')
```
