```python
def toldiruvchi(df):
    for cols in df.columns:
        if df[cols].isnull().any():
            if df[cols].dtype == 'object':
                    df[cols].fillna(df[cols].mode()[0],inplace=True)
            else:
                df[cols].fillna(df[cols].mean(),inplace=True)
    return df
```

```python
toldiruvchi(df)
```

```python
from sklearn.preprocessing import LabelEncoder
labeller = LabelEncoder()
```

```python
def encodlovchi(df):
    for cols in df.columns:
        if df[cols].dtype=='object':
            if df[cols].nunique()<=5:
                dummies=pd.get_dummies(df[cols],prefix=cols, dtype=int)
                df=pd.concat([df.drop(columns=[cols]),dummies],axis=1)
            else:
                df[cols]=labeller.fit_transform(df[cols])
    return df
```

```python
df = encodlovchi(df)
```

```python
df.sample(5)
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler()
```

```python
def scale_qiluvchi(df):
    num_col=df.select_dtypes(include=['int64','int32','float64']).columns.drop('hdlngth')
    df[num_col]=minmax.fit_transform(df[num_col])
    return df
```

```python
scale_qiluvchi(df)
```

# Linear regression

```python
x = df.drop('hdlngth', axis=1)
```

```python
y = df['hdlngth']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
linReg = LinearRegression()
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, root_mean_squared_error
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.25, random_state=63)
```

```python
linReg.fit(x_train,y_train)
```

```python
pred = linReg.predict(x_test)
```

```python
r2=r2_score(y_test,pred)
mae=mean_absolute_error(y_test,pred)
mse=mean_squared_error(y_test,pred)
rmse=root_mean_squared_error(y_test,pred)
```

```python
print(r2)
print(mae)
print(mse)
print(rmse)
```
