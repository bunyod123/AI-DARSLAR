```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\bunyo\OneDrive\Desktop\git_dataset\9. Sales-Data-Analysis.csv")
```

```python
def tozala(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='object':
                df[col].fillna(df[col].mode()[0],inplace=True)
            else:
                df[col].fillna(df[col].mean(),inplace=True)
    return df
```

```python
tozala(df)
```

```python
from sklearn.preprocessing import LabelEncoder
labeling = LabelEncoder()
```

```python
def encodla(df):
    for col in df.columns:
        if df[col].dtype=='object':
            if df[col].nunique()<=5:
                dummies=pd.get_dummies(df[col],prefix=col, dtype=int)
                df=pd.concat([df.drop(columns=[col]),dummies],axis=1)
            else:
                df[col]=labeling.fit_transform(df[col])
    return df
```

```python
df = encodla(df)
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler()
```

```python
def scale_qil(df):
    num_col=df.select_dtypes(include=['int64','int32','float64']).columns.drop('Quantity')
    df[num_col]=minmax.fit_transform(df[num_col])
    return df
```

```python
df = scale_qil(df)
```

```python
df.info()
```

# Datani Bolish

```python
x = df.drop('Quantity', axis=1)
```

```python
y = df['Quantity']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

# Linear regressiya

```python
from sklearn.linear_model import LinearRegression
lrModel = LinearRegression()
```

```python
lrModel.fit(x_train,y_train)
```

```python
pred1 = lrModel.predict(x_test)
```

```python
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
r2=r2_score(y_test,pred1)
mae=mean_absolute_error(y_test,pred1)
```

```python
print(r2)
print(mae)
```

# K fold cross validation

```python
from sklearn.model_selection import KFold, cross_val_score
```

```python
kf = KFold(n_splits=5,shuffle=True,random_state=24)
```

```python
score = cross_val_score(lrModel,X=x,y=y,cv=kf,scoring='r2')
```
