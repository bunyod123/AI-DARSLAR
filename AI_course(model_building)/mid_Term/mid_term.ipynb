```python
#1  Berilgan so’zni boshi va oxirdagi so’zlardan so’z tuzadigan funksiya yarating
def sozYasavchi(soz):
    return soz[0] + soz[-1]
```

```python
#2  Berilgan listdagi sonlar o’rta arifmetigini topadigan funskiya yarating
def meanFind(num):
    return sum(num)/len(num)
```

```python
#3  Berilgan so’zni uzunligini topadigan funksiya yarating
def sozUzunligi(soz):
    return len(soz)
```

```python
#4  Berilgan listdagi so’zlar ichidan eng uzunini toping
def sozUzun(soz):
    raqam = 0
    uzunSoz = ''
    for word in soz:
        if len(word) > raqam:
            raqam = len(word)
            uzunSoz = word
    return uzunSoz
```

```python
sozUzun(['qwqw','qwqwqwqwq','eewewewewewwewewew','112'])
```

```python
#5  Berilgan listdagi eng katta, eng kichik sonlarni topadigan funksiya yarating
def minmax(num):
    kichik = min(num)
    katta = max(num)
    return 'kattasi ' ,katta, 'kichik ' ,kichik
```

```python
minmax([1,2,3,4,5,6,7,899,1])
```

```python
#6   Funskiyaga berilgan faqat juft sonlarni kopaytmasini toping
def juftSon(nums):
    kopaytma = 1
    for num in nums:
        if num % 2 == 0:
            kopaytma *= num
    return kopaytma
```

```python
juftSon([1,2,3,4,5,6,7])
```

```python
#7 Funskiyaga berilgan faqat toq sonlarni kopaytmasini toping
def toqSon(nums):
    kopaytma = 1
    for num in nums:
        if num % 2 != 0:
            kopaytma *= num
    return kopaytma
```

```python
toqSon([1,2,3,4,5,6])
```

# Machine Learning

```python
#1 Funksiya yordamida missing qiymatlarni toldiring
def toldiruchi(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='object':
                df[col].fillna(df[col].mode()[0],inplace=True)
            else:
                df[col].fillna(df[col].mean(),inplace=True)
    return df
```

```python
#2 Funksiya yordamida encoding qiling
def encodlovchi(df):
    for col in df.columns:
        if df[col].dtype=='object':
            if df[col].nunique()<=5:
                dummies=pd.get_dummies(df[col],prefix=col, dtype=int)
                df=pd.concat([df.drop(columns=[col]),dummies],axis=1)
            else:
                df[col]=labeling.fit_transform(df[col])
    return df
```

```python
#3  Funksiya yordamida scaling  qiling
def scale_qil(df):
    num_col=df.select_dtypes(include=['int64','int32','float64']).columns.drop('Target qiymat')
    df[num_col]=minmax.fit_transform(df[num_col])
    return df
```
