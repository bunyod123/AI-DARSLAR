```python
from sklearn.preprocessing import LabelEncoder
labeller = LabelEncoder()
```

```python
def encodlovchi(df):
    for col in df.columns:
        if df[col].dtype == 'object':
            if df[col].nunique() <= 4:
                dumies = pd.get_dummies(df[col], prefix='yangi', dtype=int)
                df = pd.concat([df.drop(columns=[col]), dumies], axis=1)
            else:
                df[col] = labeller.fit_transform(df[col])
    return df
```

```python
df = encodlovchi(df)
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler()

def scale_qiluchi(df):
    num_cols = df.select_dtypes(include=['int64','int32']).columns.drop('goals')
    df[num_cols] = minmax.fit_transform(df[num_cols])
    return df
    
    
```

```python
df = scale_qiluchi(df)
```
