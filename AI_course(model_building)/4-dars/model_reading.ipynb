```python
from sklearn.preprocessing import LabelEncoder
labeller = LabelEncoder()
```

```python
def encodlovchi(df):
    for col in df.columns:
        if df[col].dtype == 'object':
            if df[col].nunique() <= 4:
                dumies = pd.get_dummies(df[col], prefix='yangi', dtype=int)
                df = pd.concat([df.drop(columns=[col]), dumies], axis=1)
            else:
                df[col] = labeller.fit_transform(df[col])
    return df
```

```python
df = encodlovchi(df)
```

```python
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler
minmax = MinMaxScaler()

def scale_qiluchi(df):
    num_cols = df.select_dtypes(include=['int64','int32']).columns.drop('goals')
    df[num_cols] = minmax.fit_transform(df[num_cols])
    return df
    
    
```

```python
df = scale_qiluchi(df)
```

# Training

```python
x = df.drop(columns='goals')
```

```python
y = df['goals']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression,LogisticRegression
from sklearn.metrics import r2_score,mean_absolute_error, mean_squared_error
lr = LinearRegression()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

# linear regression

```python
lr.fit(x_train,y_train)
```

```python
pred = lr.predict(x_test)
```

```python
r2_score(y_test,pred)
```

```python
mean_absolute_error(y_test,pred)
```

# logictik regressiya

```python
log_reg = LogisticRegression()
```

```python
log_reg.fit(x_train,y_train)
```

```python
predicton = log_reg.predict(x_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, predicton))
```
