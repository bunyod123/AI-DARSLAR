```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\bunyo\OneDrive\Desktop\git_dataset\9. Sales-Data-Analysis.csv")
```

```python
df.info()
```

```python
def tozalash(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='object':
                df[col].fillna(df[col].mode()[0],inplace=True)
            else:
                df[col].fillna(df[col].mean(),inplace=True)
    return df
```

```python
df = tozalash(df)
```

```python
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
minmax = MinMaxScaler()
labeling = LabelEncoder()
```

```python
def encodla(df):
    for col in df.columns:
        if df[col].dtype=='object':
            if df[col].nunique()<=5:
                dummies=pd.get_dummies(df[col],prefix=col, dtype=int)
                df=pd.concat([df.drop(columns=[col]),dummies],axis=1)
            else:
                df[col]=labeling.fit_transform(df[col])
    return df
```

```python
df= encodla(df)
```

```python
def scale_qil(df):
    num_col=df.select_dtypes(include=['int64','int32','float64']).columns.drop('Quantity')
    df[num_col]=minmax.fit_transform(df[num_col])
    return df
```

```python
df = scale_qil(df)
```

```python
df.info()
```

```python
from sklearn.linear_model import LinearRegression
lin_Reg = LinearRegression()
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
x = df.drop('Quantity', axis=1)
y = df['Quantity']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.25,random_state=46)
```

```python
lin_Reg.fit(x_train,y_train)
```

```python
pred1 = lin_Reg.predict(x_test)
```

```python
print(r2_score(y_test,pred1))
```

# Tree regression

```python
from sklearn.tree import DecisionTreeRegressor
tree_reg = DecisionTreeRegressor()
```

```python
tree_reg.fit(x_train,y_train)
```

```python
pred2 = tree_reg.predict(x_test)
```

```python
print(r2_score(y_test,pred2))
```
