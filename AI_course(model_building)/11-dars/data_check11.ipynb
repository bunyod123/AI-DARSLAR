```python
import pandas as pd

from sklearn.model_selection import train_test_split,KFold,cross_val_score
from sklearn.preprocessing import LabelEncoder
labeller = LabelEncoder()

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC

svc = SVC()
Log_Class = LogisticRegression()
RF_Class = RandomForestClassifier()
DT_Class = DecisionTreeClassifier()

from sklearn.metrics import classification_report
```

```python
df = pd.read_csv(r"C:\Users\bunyo\OneDrive\Desktop\git_dataset\heart.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
def encodlovchi(df):
    for cols in df.columns:
        if df[cols].dtype == 'object':
            if df[cols].nunique() >= 4:
                dum = pd.get_dummies(df[cols], prefix='new', dtype=int)
                df = pd.concat([df.drop(columns=[cols]),dum], axis=1)
            else:
                df[cols] = labeller.fit_transform(df[cols])
    return df
```

```python
df = encodlovchi(df)
```

```python
df
```

```python
x = df.drop('HeartDisease',axis=1)
y = df['HeartDisease']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=22)
```

# SVC

```python
svc.fit(x_train,y_train)
```

```python
svc_pred = svc.predict(x_test)
```

```python
print(classification_report(y_test,svc_pred))
```

# Logistic Regression

```python
Log_Class.fit(x_train,y_train)
```

```python
Log_pred = Log_Class.predict(x_test)
```

```python
print(classification_report(y_test,Log_pred))
```

# Random forest regression

```python
RF_Class.fit(x_train,y_train)
```

```python
RF_pred = RF_Class.predict(x_test)
```

```python
print(classification_report(y_test,RF_pred))
```

# Decision tree Classication

```python
DT_Class.fit(x_train,y_train)
```

```python
DT_pred = DT_Class.predict(x_test)
```

```python
print(classification_report(y_test,DT_pred))
```

# KFold orqali topish (SVC)

```python
from sklearn.model_selection import KFold,cross_val_score
import numpy as np
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=33)
crossing = cross_val_score(svc, X=x, y=y, cv=kf, scoring='accuracy')

print(np.mean(crossing))
```

# KFold orqali (Logistic Regression)

```python
crossing1 = cross_val_score(Log_Class, X=x,y=y, cv=kf, scoring='accuracy')

print(np.mean(crossing1))
```

# KFold with Random forest

```python
crossing2 = cross_val_score(RF_Class, X=x, y=y, cv=kf, scoring='accuracy')
print(np.mean(crossing2))
```

# Decision Tree with KFold

```python
crossing3 = cross_val_score(DT_Class,x, y, cv=kf, scoring='accuracy')
print(np.mean(crossing3))
```
