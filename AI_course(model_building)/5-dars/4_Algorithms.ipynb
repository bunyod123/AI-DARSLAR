```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
des_tree_class = DecisionTreeClassifier()
des_tree_reg = DecisionTreeRegressor()
from sklearn.linear_model import LinearRegression, LogisticRegression
linear_reg = LinearRegression()
log_reg = LogisticRegression()

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, r2_score, mean_absolute_error,root_mean_squared_error, mean_squared_error
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
min_max = MinMaxScaler()
labeling = LabelEncoder()
```

```python
df = pd.read_csv(r"C:\Users\bunyo\OneDrive\Desktop\git_dataset\kidney_disease_dataset.csv")
```

```python
df.head()
```

```python
for cols in df.columns:
    if df[cols].dtype == 'object':
        if df[cols].nunique() <= 3:
            dumies = pd.get_dummies(df[cols], prefix='new', dtype=int)
            pd.concat([df.drop(columns=[cols]), dumies], axis=1)
        else: 
            df[cols] = labeling.fit_transform(df[cols])
```

```python
for cols in df.columns:
    if df[cols].dtype == 'object':
        if df[cols].isnull().any():
            df[cols].fillna(df[cols].mode()[0], axis=1, inplace=True)
    else:
        df[cols].fillna(df[cols].mean(), inplace=True)
```

```python
df.info()
```

```python
x = df.drop('CKD_Status', axis=1)
y = df['CKD_Status']
```

```python
x_train,y_train,x_test,y_test = train_test_split(x,y, test_size=0.2,random_state=42)
```

# Decision tree Classifier

```python
tree_class = des_tree_class.fit(x_train,x_test)
```

```python
pred1 = tree_class.predict(y_train)
```

```python
print(classification_report(y_test, pred1))
```

# Decision tree Regressor

```python
tree_reg = des_tree_reg.fit(x_train,x_test)
```

```python
pred2 = tree_reg.predict(y_train)
```

```python
mse = mean_absolute_error(y_test, pred2)
print(mse)
```

# Linear regresion

```python
linear_reg1 = linear_reg.fit(x_train, x_test)
```

```python
mean_squared_error(y_test, linear_reg1.predict(y_train))
```

# Logistic Regression

```python
log_reg2 = log_reg.fit(x_train,x_test)
```

```python
pred3 = log_reg2.predict(y_train)
```

```python
mean_absolute_error(y_test, pred3)
```
