```python
def smash(words):
    sentence =  ""
    for word in words:
        sentence += word + " "
        
    return sentence.strip()
```

```python
def count_by(x, n):
    result = []
    we = 1
    for i in n:
        result.append(x*we)
        we += 1
    return result
```

```python
def func2(x,n):
    result = []
    for i in range(n+1):
        result.append(x*i)
    return result
```

```python
def func1(nums):
    sum = []
    min1 = min(nums)
    for i in range(len(nums)):
        if nums[i] - min1 > 0:
            sum.append(nums[i])
    min2 = min(sum)
    return min1+min2
```
